Q7.

Microsoft Power BI is a business intelligence and data visualization tool that supports many types of data sources. These data sources help businesses collect, analyze, and visualize data from different platforms in one place.

Power BI supports the following data sources:

1. File-Based Data Sources

Power BI can import data from different file formats such as:
- Excel files
- CSV files
- Text files
- XML files
- JSON files
- PDF files

These sources are commonly used for small business reporting and offline data analysis.

Example:
A company may store sales data in Excel and analyze it using Power BI.

2. Database Data Sources

Power BI supports many databases such as:
- MySQL
- Microsoft SQL Server
- Oracle Database
- PostgreSQL
- MongoDB

These databases are used for handling large amounts of business data.

Example:
An e-commerce company can connect its customer database directly to Power BI.

3. Cloud-Based Data Sources

Power BI can connect to cloud platforms and online storage services such as:
- Microsoft Azure
- Google Drive
- Dropbox
- Amazon Web Services (AWS)

These sources help businesses access and analyze cloud-stored data.

4. Online Services and Business Applications

Power BI supports integration with many online business applications such as:
- Microsoft Excel
- Salesforce
- Google Analytics
- SharePoint
- SAP

These services help organizations analyze customer data, website traffic, and business operations.

Example:
Marketing teams use Google Analytics data in Power BI to monitor website performance.

5. Web and API Data Sources

Power BI can collect data directly from:
- Websites
- REST APIs
- JSON feeds
- XML web services

These sources are useful for real-time and external data collection.

Example:
A company can import stock market data from an online API.

6. Programming and Big Data Sources

Power BI supports:
- Python
- R
- Hadoop
- Spark

These sources are mainly used for advanced analytics and machine learning.

7. Streaming and Real-Time Data Sources

Power BI can connect to live streaming data such as:
- IoT sensors
- Real-time dashboards
- Streaming datasets

These are used for live monitoring and real-time business analysis.

**Power BI supports a wide range of data sources including files, databases, cloud platforms, online services, APIs, and real-time systems. This flexibility helps businesses combine data from multiple sources and create powerful reports and dashboards for better decision-making.**

Steps to Split Owner Name in Power Query:

First Load the data and select transform option.

then step:

1. Open Power Query Editor in Power BI.

2. Select the "Owner_Name" column.

3. Go to the Transform tab.

4. Click Split Column.

5. Select By Delimiter.

6. Choose Space as the delimiter.

7. Under Split at, select Left-most delimiter.

8. Click OK.

9. Two new columns will be created:
   - Owner_Name.1 > First Name
   - Owner_Name.2 > Last Name

10. Rename the columns as:
   - First Name
   - Last Name
